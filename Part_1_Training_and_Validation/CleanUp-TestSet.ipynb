```python
import os
from PIL import Image
import imagehash

ref_dir = '/Users/kushrocks6306/Deep Learning Project/Proccessed Dataset/plantvillage_train_128x128/Tomato_Early_blight'
target_dir = '/Users/kushrocks6306/Deep Learning Project/Proccessed Dataset/Test Set/Tomato___Early_blight'

print("🔍 Computing visual hashes for Train Set...")
train_visual_hashes = set()
for f in os.listdir(ref_dir):
    if f.startswith('.'): continue
    try:
        img_path = os.path.join(ref_dir, f)
        # Calculate a visual hash (phash)
        img_hash = imagehash.phash(Image.open(img_path))
        train_visual_hashes.add(img_hash)
    except Exception as e:
        pass # Skip non-image files

print("🔍 Scanning Test Set for visual matches...")
deleted_count = 0
for f in os.listdir(target_dir):
    if f.startswith('.'): continue
    test_path = os.path.join(target_dir, f)
    
    try:
        test_hash = imagehash.phash(Image.open(test_path))
        
        # If the visual hash matches, it's the same picture
        if test_hash in train_visual_hashes:
            os.remove(test_path)
            print(f"Removed visually identical image: {f}")
            deleted_count += 1
    except Exception as e:
        pass

print(f"\n✅ Done! Successfully deleted {deleted_count} visual duplicates.")
```

```python
import os
from PIL import Image
import imagehash

# Map each potato class: [Train Folder (Reference read-only)] -> [Test Folder (Target to delete from)]
category_pairs = {
    "Potato_Early_blight": {
        "train": '/Users/kushrocks6306/Deep Learning Project/Proccessed Dataset/plantvillage_train_128x128/Potato___Early_blight',
        "test": '/Users/kushrocks6306/Deep Learning Project/Proccessed Dataset/Test Set/Potato_earlyblt' 
    },
    "Potato_healthy": {
        "train": '/Users/kushrocks6306/Deep Learning Project/Proccessed Dataset/plantvillage_train_128x128/Potato___healthy',
        "test": '/Users/kushrocks6306/Deep Learning Project/Proccessed Dataset/Test Set/Potato_healthy' 
    },
    "Potato_Late_blight": {
        "train": '/Users/kushrocks6306/Deep Learning Project/Proccessed Dataset/plantvillage_train_128x128/Potato___Late_blight',
        "test": '/Users/kushrocks6306/Deep Learning Project/Proccessed Dataset/Test Set/Potato_lateblt'
    },
}

grand_deleted_count = 0

for class_name, paths in category_pairs.items():
    ref_dir = paths["train"]   # Reference set (Train - Read-only)
    target_dir = paths["test"]  # Target set (Test - Deletions happen here)

    print(f"\n==========================================")
    print(f"📂 Processing Category: {class_name}")
    print(f"==========================================")

    if not os.path.exists(ref_dir) or not os.path.exists(target_dir):
        print(f"⚠️ Directory missing for {class_name}, skipping...")
        continue

    # 1. Compute visual hashes for Train Set
    print("🔍 Computing visual hashes for Train Set...")
    train_visual_hashes = set()
    for f in os.listdir(ref_dir):
        if f.startswith('.'): continue
        try:
            img_path = os.path.join(ref_dir, f)
            img_hash = imagehash.phash(Image.open(img_path))
            train_visual_hashes.add(img_hash)
        except Exception:
            pass

    # 2. Scan Test Set for visual matches against Train Set
    print("🔍 Scanning Test Set for visual duplicates...")
    category_deleted_count = 0
    for f in os.listdir(target_dir):
        if f.startswith('.'): continue
        test_path = os.path.join(target_dir, f)

        try:
            test_hash = imagehash.phash(Image.open(test_path))

            # Delete from Test set if it visually exists in Train set
            if test_hash in train_visual_hashes:
                os.remove(test_path)
                print(f"  ❌ Removed duplicate from Test: {f}")
                category_deleted_count += 1
        except Exception:
            pass

    print(f"✅ Class Finished: Deleted {category_deleted_count} duplicates from Test.")
    grand_deleted_count += category_deleted_count

print(f"\n🎉 ALL DEDUPLICATION COMPLETE! Total images removed from Potato Test set: {grand_deleted_count}")
```
